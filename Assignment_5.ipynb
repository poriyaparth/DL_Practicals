```python

import numpy as np
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers
from tensorflow.keras.preprocessing.text import Tokenizer

# A. Data Preparation
text = "Deep learning is a branch of machine learning that uses neural networks to understand data and recognize patterns."

tokenizer = Tokenizer()
tokenizer.fit_on_texts([text])

word_index = tokenizer.word_index
vocab_size = len(word_index) + 1

print("Vocabulary:", word_index)
print("Vocabulary Size:", vocab_size)

# B. Generate Training Data
sequence = tokenizer.texts_to_sequences([text])[0]

window_size = 2
X = []
y = []

for i in range(window_size, len(sequence) - window_size):
    context = sequence[i-2:i] + sequence[i+1:i+3]
    target = sequence[i]

    X.append(context)
    y.append(target)

X = np.array(X)
y = np.array(y)

print("\nTraining Data:", X)
print("Target Data:", y)

# C. Build and Train CBOW Model
model = keras.Sequential([
    layers.Embedding(vocab_size, 10),
    layers.GlobalAveragePooling1D(),
    layers.Dense(vocab_size, activation='softmax')
])

model.compile(
    optimizer='adam',
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)

model.fit(X, y, epochs=100, verbose=1)

# D. Output
predictions = model.predict(X, verbose=0)
predicted_words = np.argmax(predictions, axis=1)

index_word = {v: k for k, v in word_index.items()}

print("\nActual and Predicted Words:")

for actual, predicted in zip(y, predicted_words):
    print("Actual:", index_word[actual],
          "| Predicted:", index_word[predicted])
```
