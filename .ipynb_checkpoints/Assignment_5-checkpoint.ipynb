```python
import numpy as np
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers
from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences

# Sample text
text = "I love machine learning and deep learning"

# Tokenize the text
tokenizer = Tokenizer()
tokenizer.fit_on_texts([text])

# Create word index
word_index = tokenizer.word_index
vocab_size = len(word_index) + 1

print("Word Index:", word_index)
print("Vocabulary Size:", vocab_size)
```

```python
# Convert text into sequences
sequences = tokenizer.texts_to_sequences([text])[0]

# Define context window size
window_size = 2

# Generate CBOW training data
X = []
y = []

for i in range(window_size, len(sequences) - window_size):
    context = sequences[i-window_size:i] + sequences[i+1:i+window_size+1]
    target = sequences[i]

    X.append(context)
    y.append(target)

X = np.array(X)
y = np.array(y)

print("Context Words:", X)
print("Target Words:", y)
```

```python
# Define CBOW model
model = keras.Sequential([
    layers.Embedding(vocab_size, 50, input_length=4),
    layers.Lambda(lambda x: tf.reduce_mean(x, axis=1)),
    layers.Dense(vocab_size, activation='softmax')
])

# Compile the model
model.compile(
    optimizer='adam',
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)

# Train the model
history = model.fit(X, y, epochs=100, verbose=1)
```

```python
# Predict target word
predictions = model.predict(X)

# Get predicted word index
predicted_index = np.argmax(predictions[0])

# Convert index into word
reverse_word_index = {v: k for k, v in word_index.items()}

predicted_word = reverse_word_index[predicted_index]

print("Context Words:",
      [reverse_word_index[i] for i in X[0]])

print("Actual Target Word:",
      reverse_word_index[y[0]])

print("Predicted Target Word:", predicted_word)
```

```python
# ============================================================
# CBOW (Continuous Bag of Words) Word Embedding Model
# Predict the target (middle) word from its surrounding context words
# ============================================================

import numpy as np
import matplotlib.pyplot as plt
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers
from tensorflow.keras.preprocessing.text import Tokenizer

np.random.seed(42)
tf.random.set_seed(42)

# ---------- 1. Sample text (several sentences give more training data) ----------
corpus = [
    "I love machine learning and deep learning",
    "deep learning is a part of machine learning",
    "machine learning helps computers learn from data",
    "I love learning new things about deep learning",
    "neural networks are used in deep learning",
    "computers learn patterns from data using machine learning",
]

# ---------- 2. Tokenize ----------
tokenizer = Tokenizer()
tokenizer.fit_on_texts(corpus)

word_index = tokenizer.word_index
reverse_word_index = {v: k for k, v in word_index.items()}
vocab_size = len(word_index) + 1          # +1 because index 0 is reserved

print("Word Index:", word_index)
print("Vocabulary Size:", vocab_size)

sequences = tokenizer.texts_to_sequences(corpus)

# ---------- 3. Generate CBOW training data ----------
window_size = 2                           # 2 words left + 2 words right
X, y = [], []

for seq in sequences:
    for i in range(window_size, len(seq) - window_size):
        context = seq[i - window_size:i] + seq[i + 1:i + window_size + 1]
        target = seq[i]
        X.append(context)
        y.append(target)

X = np.array(X)
y = np.array(y)

print("\nTotal training samples:", len(X))
print("Sample context :", [reverse_word_index[i] for i in X[0]])
print("Sample target  :", reverse_word_index[y[0]])

# ---------- 4. Build the CBOW model ----------
embedding_dim = 50

model = keras.Sequential([
    layers.Input(shape=(2 * window_size,)),             # 4 context words
    layers.Embedding(vocab_size, embedding_dim),        # word -> dense vector
    layers.GlobalAveragePooling1D(),                    # average context vectors
    layers.Dense(vocab_size, activation="softmax"),     # probability of each word
])

model.summary()

# ---------- 5. Compile ----------
model.compile(
    optimizer="adam",
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"],
)

# ---------- 6. Train ----------
history = model.fit(X, y, epochs=200, batch_size=8, verbose=0)
print("\nFinal loss    :", round(history.history["loss"][-1], 4))
print("Final accuracy:", round(history.history["accuracy"][-1], 4))

plt.figure(figsize=(9, 3.5))
plt.subplot(1, 2, 1)
plt.plot(history.history["loss"])
plt.title("Training Loss"); plt.xlabel("Epoch")
plt.subplot(1, 2, 2)
plt.plot(history.history["accuracy"])
plt.title("Training Accuracy"); plt.xlabel("Epoch")
plt.tight_layout()
plt.show()

# ---------- 7. Predict target words ----------
predictions = model.predict(X, verbose=0)

print("\n--- Predictions (first 5 samples) ---")
for i in range(min(5, len(X))):
    pred_idx = np.argmax(predictions[i])
    print("Context:", [reverse_word_index[w] for w in X[i]],
          "| Actual:", reverse_word_index[y[i]],
          "| Predicted:", reverse_word_index[pred_idx])

# Predict for your own context words
def predict_word(context_words):
    idx = [word_index[w] for w in context_words]
    probs = model.predict(np.array([idx]), verbose=0)[0]
    return reverse_word_index[int(np.argmax(probs))]

print("\nCustom test:", ["machine", "learning", "and", "learning"],
      "->", predict_word(["machine", "learning", "and", "learning"]))

# ---------- 8. Learned word embeddings ----------
embeddings = model.layers[0].get_weights()[0]       # shape (vocab_size, 50)
print("\nEmbedding matrix shape:", embeddings.shape)

for word in ["machine", "learning", "deep"]:
    print(word, "->", np.round(embeddings[word_index[word]][:5], 3), "...")

# Cosine similarity between two words
def similarity(w1, w2):
    a, b = embeddings[word_index[w1]], embeddings[word_index[w2]]
    return float(np.dot(a, b) / (np.linalg.norm(a) * np.linalg.norm(b)))

print("\nSimilarity(machine, deep):", round(similarity("machine", "deep"), 3))
print("Similarity(machine, data):", round(similarity("machine", "data"), 3))
```
